```python
import json
import os
import time
from pathlib import Path
from sentence_transformers import SentenceTransformer, util
import numpy as np
from openai import OpenAI


model = SentenceTransformer('multi-qa-MiniLM-L6-cos-v1')

API_KEY_PATH = Path('openai_key.txt')

```

```python
def load_openai_api_key():
    env_key = os.environ.get('OPENAI_API_KEY')
    if env_key:
        return env_key.strip()

    if API_KEY_PATH.exists():
        file_key = API_KEY_PATH.read_text().strip()
        if file_key:
            return file_key

    raise EnvironmentError('Set OPENAI_API_KEY or create openai_key.txt in the notebook directory.')


def generate_no_ad_outputs(prompt: str, n: int = 10, model_name: str = 'gpt-4-turbo', temperature: float = 1.0, sleep_seconds: float = 0.5):

    client = OpenAI(api_key=load_openai_api_key())
    outputs = []

    for _ in range(n):
        response = client.chat.completions.create(
            model=model_name,
            temperature=temperature,
            max_tokens=500,
            messages=[
                {'role': 'user', 'content': prompt},
                {'role': 'assistant', 'content': ''},
            ],
        )
        outputs.append(response.choices[0].message.content)
        time.sleep(sleep_seconds)

    return outputs




def load_saved_outputs(path: Path = OUTPUT_PATH):
    payload = json.loads(path.read_text())
    return payload['outputs'], payload
```

# Scenario 1

```python

prompt = '"What can I visit on a trip to Hawaii?"'

OUTPUT_PATH = Path('generated_no_ad_outputs_scenario_1.json')


def save_outputs(outputs: list[str], path: Path = OUTPUT_PATH, prompt: str = prompt, model_name: str = 'gpt-4-turbo', temperature: float = 1.0):
    payload = {
        'prompt': prompt,
        'model': model_name,
        'temperature': temperature,
        'count': len(outputs),
        'outputs': outputs,
    }
    path.write_text(json.dumps(payload, indent=2, ensure_ascii=False))
    return payload
```

```python
query = f'''
{prompt}
Answer this question for 3 sentences.
'''
```

```python

OUTPUT_PATH = Path('generated_no_ad_outputs_scenario_1.json')

no_ad_outputs = generate_no_ad_outputs(query, n=10)
payload = save_outputs(no_ad_outputs)

print(f"Saved {payload['count']} outputs to {OUTPUT_PATH.resolve()}")
for i, text in enumerate(no_ad_outputs, start=1):
    print(f'{i}. {text}\n')
```
